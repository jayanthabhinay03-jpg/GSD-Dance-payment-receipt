```python
# =========================================================
#                  GSD DANCE STUDIO
#                   PAYMENT RECEIPT
# =========================================================

# Student Details
student_name = input("Enter Student Name: ")
age = int(input("Enter Age: "))
dance_type = input("Enter Dance Type: ")
session = input("Enter Session (Morning/Evening): ")

# Payment Details
monthly_fee = 1000

advance_payment = float(
    input("Enter Advance Payment: ₹")
)

payment_mode = input("Enter Payment Mode (Cash/UPI): ")

# Calculations
remaining_balance = monthly_fee - advance_payment

# Receipt Number
receipt_number = "GSD-001"

# Display Receipt
print("\n")
print("╔" + "═" * 54 + "╗")
print("║" + " " * 54 + "║")
print("║" + "              GSD DANCE STUDIO".center(54) + "║")
print("║" + "           DANCE • FITNESS • FUN".center(54) + "║")
print("║" + " " * 54 + "║")
print("║" + "              PAYMENT RECEIPT".center(54) + "║")
print("║" + " " * 54 + "║")
print("╠" + "═" * 54 + "╣")

print("║ Receipt No. :", receipt_number.ljust(15),
      "Date : 05-10-2026 ║")

print("╠" + "═" * 54 + "╣")
print("║" + " " * 54 + "║")
print("║" + "              STUDENT DETAILS".center(54) + "║")
print("║" + " " * 54 + "║")
print("╠" + "═" * 54 + "╣")

print("║ Student Name   :", student_name.ljust(35) + "║")
print("║ Age            :", str(age).ljust(35) + "║")
print("║ Dance Type     :", dance_type.ljust(35) + "║")
print("║ Session        :", session.ljust(35) + "║")

print("╠" + "═" * 54 + "╣")
print("║" + " " * 54 + "║")
print("║" + "              PAYMENT DETAILS".center(54) + "║")
print("║" + " " * 54 + "║")
print("╠" + "═" * 54 + "╣")

print("║ Monthly Fee          : ₹", format(monthly_fee, ".2f").ljust(27) + "║")
print("║ Advance Payment      : ₹", format(advance_payment, ".2f").ljust(27) + "║")
print("║ Remaining Balance    : ₹", format(remaining_balance, ".2f").ljust(27) + "║")
print("║ Payment Mode         :", payment_mode.ljust(31) + "║")

print("╠" + "═" * 54 + "╣")

# Payment Status
if remaining_balance == 0:
    status = "PAID"
else:
    status = "PARTIALLY PAID"

print("║ Payment Status       :", status.ljust(31) + "║")

print("╠" + "═" * 54 + "╣")

# Loop for footer message
for i in range(2):
    print("║" + "       Thank You for Choosing GSD Dance Studio!".center(54) + "║")

print("║" + " " * 54 + "║")
print("║ Student Signature: ___________".ljust(55) + "║")
print("║ Authorized Signature: ________".ljust(55) + "║")
print("║" + " " * 54 + "║")

print("╚" + "═" * 54 + "╝")
```
